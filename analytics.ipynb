```python
%sql
CREATE SCHEMA IF NOT EXISTS healthcare_catalog.analytics;
```

```python
%sql
SHOW SCHEMAS IN healthcare_catalog;
```

```python
%sql
--KPI 1 — Total Patients..
SELECT
    COUNT(*) AS total_patients
FROM healthcare_catalog.gold.dim_patient;
```

```python
%sql 
--KPI 2 — Total Encounters
SELECT
    COUNT(*) AS total_encounters
FROM healthcare_catalog.gold.fact_encounter;
```

```python
%sql
--KPI 3 — Encounters by Type
SELECT
    encounter_class,
    COUNT(*) AS total_encounters
FROM healthcare_catalog.gold.fact_encounter
GROUP BY encounter_class
ORDER BY total_encounters DESC;
```

```python
%sql
--KPI 4 — Total Healthcare Cost
SELECT
    ROUND(SUM(total_claim_cost), 2) AS total_healthcare_cost
FROM healthcare_catalog.gold.fact_encounter;
```

```python
%sql

-- KPI 5 — Average Encounter Cost
SELECT
    ROUND(AVG(total_claim_cost), 2) AS avg_encounter_cost
FROM healthcare_catalog.gold.fact_encounter;
```

```python
%sql
--KPI 6 — Healthcare Cost by Encounter Type
SELECT
    encounter_class,
    COUNT(*) AS encounters,
    ROUND(SUM(total_claim_cost), 2) AS total_cost,
    ROUND(AVG(total_claim_cost), 2) AS average_cost
FROM healthcare_catalog.gold.fact_encounter
GROUP BY encounter_class
ORDER BY total_cost DESC;
```

```python
%sql
--KPI 7 — Top Conditions
SELECT
    condition_description,
    COUNT(*) AS condition_count
FROM healthcare_catalog.gold.fact_condition
GROUP BY condition_description
ORDER BY condition_count DESC
LIMIT 10;
```

```python
%sql
--KPI 8 — Top Medications
SELECT
    medication_description,
    COUNT(*) AS prescription_count
FROM healthcare_catalog.gold.fact_medication
GROUP BY medication_description
ORDER BY prescription_count DESC
LIMIT 10;
```

```python
%sql
--KPI 9 — Top Procedures
SELECT
    procedure_description,
    COUNT(*) AS procedure_count
FROM healthcare_catalog.gold.fact_procedure
GROUP BY procedure_description
ORDER BY procedure_count DESC
LIMIT 10;
```

```python
%sql
--KPI 10 — Patient Healthcare Expenses
SELECT
    patient_id,
    first_name,
    last_name,
    healthcare_expenses
FROM healthcare_catalog.gold.dim_patient
ORDER BY healthcare_expenses DESC
LIMIT 10;
```

```python
%sql
-- KPI 11 — Patient Encounter Count
SELECT
    patient_id,
    COUNT(*) AS encounter_count
FROM healthcare_catalog.gold.fact_encounter
GROUP BY patient_id
ORDER BY encounter_count DESC
LIMIT 10;
```

```python
%sql
--KPI 12 — Monthly Encounters
SELECT
    d.year,
    d.month,
    d.month_name,
    COUNT(e.encounter_id) AS total_encounters
FROM healthcare_catalog.gold.fact_encounter e

JOIN healthcare_catalog.gold.dim_date d
    ON CAST(date_format(e.start_datetime, 'yyyyMMdd') AS INT)
       = d.date_key

GROUP BY
    d.year,
    d.month,
    d.month_name

ORDER BY
    d.year,
    d.month;
```

```python
%sql
--KPI 13 — Encounters by Year
SELECT
    d.year,
    COUNT(e.encounter_id) AS total_encounters
FROM healthcare_catalog.gold.fact_encounter e

JOIN healthcare_catalog.gold.dim_date d
    ON CAST(date_format(e.start_datetime, 'yyyyMMdd') AS INT)
       = d.date_key

GROUP BY d.year
ORDER BY d.year;
```

```python
%sql
--Create an Analytics KPI Table
CREATE OR REPLACE TABLE healthcare_catalog.analytics.healthcare_kpis
AS

SELECT
    'Total Patients' AS kpi_name,
    CAST(COUNT(*) AS DOUBLE) AS kpi_value
FROM healthcare_catalog.gold.dim_patient

UNION ALL

SELECT
    'Total Encounters',
    CAST(COUNT(*) AS DOUBLE)
FROM healthcare_catalog.gold.fact_encounter

UNION ALL

SELECT
    'Total Conditions',
    CAST(COUNT(*) AS DOUBLE)
FROM healthcare_catalog.gold.fact_condition

UNION ALL

SELECT
    'Total Medications',
    CAST(COUNT(*) AS DOUBLE)
FROM healthcare_catalog.gold.fact_medication

UNION ALL

SELECT
    'Total Procedures',
    CAST(COUNT(*) AS DOUBLE)
FROM healthcare_catalog.gold.fact_procedure

UNION ALL

SELECT
    'Total Immunizations',
    CAST(COUNT(*) AS DOUBLE)
FROM healthcare_catalog.gold.fact_immunization

UNION ALL

SELECT
    'Total Healthcare Cost',
    ROUND(
        COALESCE(SUM(total_claim_cost), 0),
        2
    )
FROM healthcare_catalog.gold.fact_encounter;
```

```python
%sql
SELECT *
FROM healthcare_catalog.analytics.healthcare_kpis;
```

```python
%sql
--Create Monthly Encounter Analytics
CREATE OR REPLACE TABLE healthcare_catalog.analytics.monthly_encounters
AS

SELECT
    d.year,
    d.month,
    d.month_name,
    COUNT(e.encounter_id) AS total_encounters,
    ROUND(
        SUM(e.total_claim_cost),
        2
    ) AS total_cost,
    ROUND(
        AVG(e.total_claim_cost),
        2
    ) AS average_cost

FROM healthcare_catalog.gold.fact_encounter e

JOIN healthcare_catalog.gold.dim_date d
    ON CAST(
        date_format(e.start_datetime, 'yyyyMMdd')
        AS INT
    ) = d.date_key

GROUP BY
    d.year,
    d.month,
    d.month_name;
```

```python
%sql
SELECT *
FROM healthcare_catalog.analytics.monthly_encounters
ORDER BY year, month;
```

```python
%sql
--Create Condition Analytics
CREATE OR REPLACE TABLE healthcare_catalog.analytics.top_conditions
AS

SELECT
    condition_code,
    condition_description,
    COUNT(*) AS condition_count

FROM healthcare_catalog.gold.fact_condition

GROUP BY
    condition_code,
    condition_description

ORDER BY condition_count DESC;
```

```python
%sql
--Create Medication Analytics
CREATE OR REPLACE TABLE healthcare_catalog.analytics.top_medications
AS

SELECT
    medication_code,
    medication_description,
    COUNT(*) AS medication_count,
    ROUND(SUM(total_cost), 2) AS total_medication_cost

FROM healthcare_catalog.gold.fact_medication

GROUP BY
    medication_code,
    medication_description

ORDER BY medication_count DESC;
```
