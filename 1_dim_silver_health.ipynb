```python
#Import libraries
from pyspark.sql.functions import (
    col,
    trim,
    upper,
    lower,
    regexp_replace,
    to_date,
    when,
    lit,
    current_timestamp
)
```

```python
#Read Bronze Patients
patients_bronze = spark.table(
    "healthcare_catalog.bronze.brz_patients"
)

display(patients_bronze)
```

```python
#Select required columns
patients_silver = patients_bronze.select(
    "Id",
    "BIRTHDATE",
    "DEATHDATE",
    "FIRST",
    "MIDDLE",
    "LAST",
    "MARITAL",
    "RACE",
    "ETHNICITY",
    "GENDER",
    "CITY",
    "STATE",
    "COUNTY",
    "ZIP",
    "LAT",
    "LON",
    "HEALTHCARE_EXPENSES",
    "HEALTHCARE_COVERAGE",
    "INCOME"
)
```

```python

#Rename columns

patients_silver = patients_silver \
    .withColumnRenamed("Id", "patient_id") \
    .withColumnRenamed("BIRTHDATE", "birth_date") \
    .withColumnRenamed("DEATHDATE", "death_date") \
    .withColumnRenamed("FIRST", "first_name") \
    .withColumnRenamed("MIDDLE", "middle_name") \
    .withColumnRenamed("LAST", "last_name") \
    .withColumnRenamed("MARITAL", "marital_status") \
    .withColumnRenamed("RACE", "race") \
    .withColumnRenamed("ETHNICITY", "ethnicity") \
    .withColumnRenamed("GENDER", "gender") \
    .withColumnRenamed("CITY", "city") \
    .withColumnRenamed("STATE", "state") \
    .withColumnRenamed("COUNTY", "county") \
    .withColumnRenamed("ZIP", "zip_code") \
    .withColumnRenamed("LAT", "latitude") \
    .withColumnRenamed("LON", "longitude") \
    .withColumnRenamed(
        "HEALTHCARE_EXPENSES",
        "healthcare_expenses"
    ) \
    .withColumnRenamed(
        "HEALTHCARE_COVERAGE",
        "healthcare_coverage"
    )
```

```python
#Standardize text
#gender
patients_silver = patients_silver.withColumn(
    "gender",
    upper(trim(col("gender")))
)
#race
patients_silver = patients_silver.withColumn(
    "race",
    trim(col("race"))
)
#state
patients_silver = patients_silver.withColumn(
    "state",
    upper(trim(col("state")))
)
```

```python
#Remove invalid patient IDs
patients_silver = patients_silver.filter(
    col("patient_id").isNotNull()
)
```

```python
# remove blank IDs:
patients_silver = patients_silver.filter(
    trim(col("patient_id")) != ""
)
```

```python
#Remove duplicate patients
patients_silver = patients_silver.dropDuplicates(
    ["patient_id"]
)
```

```python
display(patients_silver)
```

```python
#Add Silver metadata
patients_silver = patients_silver.withColumn(
    "silver_processed_timestamp",
    current_timestamp()
)
```

```python
patients_silver = patients_silver.join(
    patients_bronze.select(
        "Id",
        "_batch_id",
        "_source_file",
        "_ingest_timestamp"
    ),
    patients_silver.patient_id ==
    patients_bronze.Id,
    "left"
).drop("Id")
```

```python
patients_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "healthcare_catalog.silver.slv_patients"
    )
```

```python
%sql
--Conditions without patients..

SELECT COUNT(*) AS orphan_conditions
FROM healthcare_catalog.silver.slv_conditions c
LEFT JOIN healthcare_catalog.silver.slv_patients p
    ON c.patient_id = p.patient_id
WHERE p.patient_id IS NULL;
```

```python
%sql
--Medications without patients
SELECT COUNT(*) AS orphan_medications
FROM healthcare_catalog.silver.slv_medications m
LEFT JOIN healthcare_catalog.silver.slv_patients p
    ON m.patient_id = p.patient_id
WHERE p.patient_id IS NULL;
```

```python
%sql
--Observations without patients
SELECT COUNT(*) AS orphan_observations
FROM healthcare_catalog.silver.slv_observations o
LEFT JOIN healthcare_catalog.silver.slv_patients p
    ON o.patient_id = p.patient_id
WHERE p.patient_id IS NULL;
```

```python
%sql
--Check encounter relationships
SELECT COUNT(*) AS orphan_encounters
FROM healthcare_catalog.silver.slv_encounters e
LEFT JOIN healthcare_catalog.silver.slv_patients p
    ON e.patient_id = p.patient_id
WHERE p.patient_id IS NULL;
```

```python
%sql
--Creating a Silver data-quality summary...

CREATE OR REPLACE TABLE healthcare_catalog.silver.silver_quality_summary
AS

SELECT
    'patients' AS table_name,
    COUNT(*) AS total_records,
    COUNT(patient_id) AS valid_records,
    COUNT(*) - COUNT(patient_id) AS null_records
FROM healthcare_catalog.silver.slv_patients

UNION ALL

SELECT
    'encounters',
    COUNT(*),
    COUNT(encounter_id),
    COUNT(*) - COUNT(encounter_id)
FROM healthcare_catalog.silver.slv_encounters

UNION ALL

SELECT
    'conditions',
    COUNT(*),
    COUNT(condition_code),
    COUNT(*) - COUNT(condition_code)
FROM healthcare_catalog.silver.slv_conditions

UNION ALL

SELECT
    'medications',
    COUNT(*),
    COUNT(medication_code),
    COUNT(*) - COUNT(medication_code)
FROM healthcare_catalog.silver.slv_medications

UNION ALL

SELECT
    'observations',
    COUNT(*),
    COUNT(observation_code),
    COUNT(*) - COUNT(observation_code)
FROM healthcare_catalog.silver.slv_observations

UNION ALL

SELECT
    'procedures',
    COUNT(*),
    COUNT(procedure_code),
    COUNT(*) - COUNT(procedure_code)
FROM healthcare_catalog.silver.slv_procedures

UNION ALL

SELECT
    'allergies',
    COUNT(*),
    COUNT(allergy_code),
    COUNT(*) - COUNT(allergy_code)
FROM healthcare_catalog.silver.slv_allergies

UNION ALL

SELECT
    'immunizations',
    COUNT(*),
    COUNT(immunization_code),
    COUNT(*) - COUNT(immunization_code)
FROM healthcare_catalog.silver.slv_immunizations

UNION ALL

SELECT
    'careplans',
    COUNT(*),
    COUNT(careplan_id),
    COUNT(*) - COUNT(careplan_id)
FROM healthcare_catalog.silver.slv_careplans;
```
